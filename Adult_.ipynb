```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/cleaned_adult.csv")
df.head()
```

```python
df.drop('Unnamed: 0', axis =1,inplace = True)
df.head()
```

```python
##  Finding and removing Outlier


df.boxplot('Age')
```

```python
## Box plot is based on Inter Quantile Range (IQR)

Q1 = df['Age'].quantile(.25)
Q3 = df['Age'].quantile(.75)

IQR = Q3-Q1

LL = Q1 - 1.5*IQR
UL = Q3 + 1.5*IQR

median = df['Age'].median()
df['Age'] = np.where(df['Age']>UL , median,df['Age'])
df['Age'] = np.where(df['Age']<LL,median,df['Age'])
```

```python
df.boxplot('Age')
```

```python
### Analysis


# filtering, Groupby, sorting, limit, pivottable
```

```python
## Filter

df[df['Age']>50]  # Single filter

df[(df['Age']>50) & (df['Race'] == 'White')]   # Multiple Filter  (& is for 'and', | is for 'or')
```

```python
## Sorting


df[(df['Age']>50) & (df['Race'] == 'White')].sort_values(by = 'Age')  # For Ascending order

df[(df['Age']>50) & (df['Race'] == 'White')].sort_values(by = 'Age',ascending = False)   # For Descending Order
```

```python
## Filter records where Occupation is "Exec-managerial"or "Prof-speciality". Sort the result first by occupation, then by age in descending order

df[(df['Occupation']=='Exec-managerial') | (df['Occupation'] == 'Prof-speciality')].sort_values(by =['Occupation','Age'],ascending =[True , False])
```

```python
## Filter Female records where "Maritial Status" is "Never-married". Sort the result by Education in descending order, then by Age in ascending order

df[df['Maritial'] == 'Never-married'].sort_values(by = ['Num','Age'], ascending=[False,True])
```

```python
## Group By


df.groupby('Workclass')['Age'].mean()    ## Single column group by

df.groupby(['Workclass','Gender'])['Hoursperweek'].mean()  ## Multiple column groupby

df.groupby('Gender')[['Age','Hoursperweek']].mean()       ## Multiple column groupby

df.groupby(['Workclass','Gender'])['Hoursperweek'].mean().unstack()   ##Converting it to matrix form using unstack
```

```python
## Group the dataset by income and find the average capital-gain and capital-loss for each income category


df.groupby('Income')[['Gains','Loss']].mean().round(2)
```

```python
## Group the dataset by both education and income ,then count how many people fall into each combination.


df.groupby(['Education','Income'])['Income'].count().unstack()
```

```python
## Group the dataset by Workclass and find the maximum hoursperweek worked in each class


df.groupby('Workclass')['Hoursperweek'].max()
```

```python
## Pivot Table


pd.pivot_table(columns = ['Workclass','Income'], index = ['Occupation'], values = ['Age'], aggfunc = ['mean'], data = df)
```

```python
## Create a pivot table of average hoursperweek by education and income.


pd.pivot_table(columns = ['Education'], index = ['Income'], values = ['Hoursperweek'], aggfunc = ['mean'], data = df)
```

```python
## Create a pivot table showing count of people by maritial-status and income

pd.pivot_table(columns = ['Income'], index = ['Maritial'], values = ['Age'], aggfunc = ['count'], data = df).stack()
```

```python
## 1.  Create age group like 18-25, 26-35, 36-45, 46-55, 66+.For each age group, Calculate the percentage of people earning >50k.


bins = [17,25,35,45,55,65,100]
labels = ['18-25', '26-35', '36-45', '46-55', '56-65', '65+']
df['Age_group'] = pd.cut(df.Age,bins = bins, labels = labels)
df.groupby('Age_group')['Income'].apply(lambda x: (x == ">50K").mean()*100).plot(kind ='bar')
```

```python
age_income_summary = df.groupby('Age_group').agg(  total_people =('Income','count'),above_50K =('Income',lambda x:(x == "50K").sum())).plot(kind = 'bar')
```

```python
## 2. Finding top 10 occupation were percentage of people earning >50k is highest. Ignore occupation with fewer than 100 records.

df['Occupation'].value_counts()
```

```python

df.groupby('Occupation')['Income'].apply(lambda x:(x == ">50K").mean()*100).sort_values(ascending = False).head(10).plot(kind ='barh')
```

```python
var = df.groupby('Occupation').agg(total_records = ('Income','count'),above_50K = ('Income',lambda x:(x == ">50K").mean()*100))
var[var['total_records'] >100]['above_50K'].sort_values(ascending = False).head(10).plot(kind ='barh')
```

```python
## 3. Comparing the average capital-gain across  different education level and visualize the top 10 education categories.

df.groupby('Education')['Gains'].mean().sort_values(ascending = False).head(10).plot(kind ='bar')
```

```python
## 4. Separate Histogram for hour-per-week for people earning <=50k and >50k.

df[df['Income'] == '<=50K']['Hoursperweek'].plot(kind = 'hist')
df[df['Income'] == '>50K']['Hoursperweek'].plot(kind = 'hist')
```

```python
## Finding the Top 10 native countries with the highest percentage of people earning >50k. Consider only countries with at least 50 records.

var = df.groupby('Country').agg(total_records = ('Income','count'),above_50K = ('Income',lambda x:(x == ">50K").mean()*100))
var[var['total_records'] >50]['above_50K'].sort_values(ascending = False).head(10).plot(kind ='barh')
```

```python
## 6. Findig the average hours-per-week for each maritial status category and visualising it from highest to lowest.

df.groupby('Maritial')['Hoursperweek'].mean().sort_values(ascending = False).head(10).plot(kind ='bar')
```

```python
## 7. For each relationship category,show the count of people earning >50k and <=50k using a stacked bar chart.

pd.crosstab(df['Relationship'],df['Income']).plot(kind = 'bar')
```

```python
## 8. For each race category, calculating the percentage of people earning >50k .

df.groupby('Race')['Income'].apply(lambda x:(x == ">50K").mean()*100).sort_values().plot(kind ='bar')
```
